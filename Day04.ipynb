```python
numbers = [1, 2, 3, 4, 5]
k = 2

def rotate(n, k):
  if not numbers:
    return []

  k %= len(n)

  if k == 0:
    return n.copy()

  return n[-k:] + n[:-k]

print(rotate(numbers, k))
```

```python
# 2: Merge Two Inventory Dictionaries
inventory1 = {
    "apple": 10,
    "banana": 5,
    "mango": 8
}

inventory2 = {
    "banana": 3,
    "orange": 6,
    "mango": 2
}

c = inventory1.copy()
for p, q in inventory2.items():
  if p in c:
    c[p] += q
  else:
    c[p] = q
c
```

```python
# 3: Calculate the Sum of Digits Using Recursion
number = 5831
def calcSum(n):
  s = 0
  while(n>0):
    temp = n % 10
    s += temp
    n = n // 10
  return s
print(calcSum(number))

```

```python
n = 5831
def calcSum(n):
  if n < 10:
    return n
  return (n%10) + calcSum(n//10)
print(calcSum(n))
```

```python
from decorator import append
# 4: Music Playlist Manager
class Playlist:
  def __init__(self, name):
    self.name = name
    self.songs = []

  def add_song(self, song):
    if song in self.songs:
      print(f"{song} already exist in playlist!")
    else:
      self.songs.append(song)
      print(f"Added: {song}")

  def remove_song(self, song):
    if song in self.songs:
      self.songs.remove(song)
      print(f"Removed: {song}")
    else:
      print(f"{song} not present in playlist.")

  def display_playlist(self):
    print(f"\n Playlist: {self.name}")

    if not self.songs:
      print("No songs in playlist.")
      return

    for index, song in enumerate(self.songs, start=1):
      print(f"{index}. {song}")

  def song_count(self):
    return len(self.songs)

playlist1 = Playlist('Workout Mix')
playlist2 = Playlist('Relaxing Music')

playlist1.add_song("Believer")
playlist1.add_song("Thunder")
playlist1.add_song("Night Changes")
playlist1.add_song("Believer")

playlist1.display_playlist()
print('Total songs:', playlist1.song_count())

playlist1.remove_song("Believer")
playlist1.display_playlist()
print('Total songs:', playlist1.song_count())

playlist2.add_song("Perfect")

playlist2.display_playlist()
playlist1.display_playlist()

```

```python
# 5: Analyze Store Sales Using NumPy
import numpy as np

sales = np.array([
    [12, 8, 15],
    [10, 14, 9],
    [7, 11, 13]
])

products = ["Product A", "Product B", "Product C"]
stores = ["Store 1", "Store 2", "Store 3"]

store_totals = np.sum(sales, axis=1)
product_totals = np.sum(sales, axis = 0)

print("Sales per store:")
for i, total in enumerate(store_totals):
  print(f"{stores[i]}: {total}")

for i, total in enumerate(product_totals):
  print(f"{products[i]}: {total}")

best_product = np.argmax(product_totals)
best_store = np.argmax(store_totals)

print("Best product:", products[best_product])
print("Best store:", stores[best_store])

print('Overall sales:', np.sum(sales))
```
