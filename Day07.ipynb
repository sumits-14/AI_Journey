```python
# 1: Find All Pairs That Sum to a Target
numbers = [2, 7, 4, 5, 3, 8, 1, 6]
target = 9
res = []
for i in range(len(numbers)):
  for j in range(1, len(numbers)):
    if numbers[i] + numbers[j] == target:
      if tuple([numbers[i], numbers[j]]) not in res and tuple([numbers[j], numbers[i]]) not in res:
        res.append(tuple([numbers[i], numbers[j]]))
print(res)
```

```python
# 2: Convert a Nested List into a Flat List
nested = [1, [2, 3], [4, [5, 6]], 7]
res = []
def sep(l):
  for i in l:
    if type(i) == int:
      res.append(i)
    elif type(i) == list:
      sep(i)

for i in nested:
  if type(i) == int:
    res.append(i)
  elif type(i) == list:
    sep(i)

res
```

```python
# 3: Build a Custom Text Encoder
text = "PythonZz 3.0!"
res = ''
for i in range(len(text)):
  if text[i].isalpha():
    temp = ord(text[i])
    if temp == 90:
      temp = 64
    elif temp == 122:
      temp = 96
    res += chr(temp+1)
  else:
    res+=text[i]

res
```

```python
# 4: Quiz Application System (OOP)
class Quiz:
  def __init__(self):
    self.questions = []
    self.current_index = 0
    self.score = 0

  def add_question(self, question, options, correct_answer):
    self.questions.append({
        'question' : question,
        'options' : options,
        'correct_answer' : correct_answer
    })

  def display_question(self):
    if self.current_index >= len(self.questions):
      print('Quize completed!')
      return

    question = self.questions[self.current_index]
    print(f"\n Q.{self.current_index + 1}: {question['question']}")

    for index, option in enumerate(question['options'], start=1):
      print(f"{index}. {option}")

  def submit_answer(self, answer):
    if self.current_index >= len(self.questions):
      print('No more questions!')
      return

    question = self.questions[self.current_index]

    if answer.strip().lower() == question['correct_answer'].strip().lower():
      self.score += 1
      print("Correct answer!")
    else:
        print("Incorrect answer!")

    self.current_index += 1

  def show_score(self):
      total = len(self.questions)
      print(f"Final score: {self.score}/{total}")


quiz = Quiz()

quiz.add_question(
    "Which keyword defines a function in Python?",
    ["func", "def", "function", "method"],
    "def"
)

quiz.add_question(
    "Which data type stores key-value pairs?",
    ["List", "Tuple", "Dictionary", "Set"],
    "Dictionary"
)

quiz.display_question()
quiz.submit_answer("def")

quiz.display_question()
quiz.submit_answer("List")

quiz.show_score()
```

```python
# 5: Analyze and Clean Missing Data Using Pandas
import pandas as pd

data = {
    "Employee": ["Amit", "Priya", "Rahul", "Sneha", "Vijay"],
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [50000, 60000, None, 55000, None]
}

df = pd.DataFrame(data)
# missing_values = pd.isna(df['Salary'])
missing_values = df.isna().sum()
print('Missing Values:')
print(missing_values)
# print(df)
dm = df.groupby('Department')['Salary'].transform('mean')
# print(dm)

df['Salary'] = df['Salary'].fillna(dm)
print()
print('Cleaned dataframe:')
print(df)
print()
print('Mean Salary:')
print(df.groupby('Department')['Salary'].mean())
print()
print("Employee's with salary > 55000:")
print(df.loc[df['Salary'] > 55000, 'Employee'].to_list())
print()
print('Employee with highest salary:')
print(df.loc[df['Salary'].idxmax(), 'Employee'])
print('Salary:', df.loc[df['Salary'].idxmax(), 'Salary'])
```
