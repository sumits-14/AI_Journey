```python
# 2: Group Words by Their Length
words = ["apple", "bat", "cat", "dog", "banana", "kiwi"]
fr = {}
for i in words:
  if len(i) in fr:
    fr[len(i)].append(i)
  else:
    fr[len(i)] = [i]
print(fr)
```

```python
# 3: Validate Balanced Parentheses
# expression = "{[()]}"
expression = "{[(])}"
# expression = "((()))"
# expression = "(()"
stack = []
if len(expression)%2==0:
  for i in expression:
    if i == '{' or i == '[' or i == '(':
      stack.append(i)

    elif i == '}' or i == ']' or i == ')':
      top = stack.pop()

      if i == ')' and top != '(':
        print('Not Balanced!')
        break

      if i == ']' and top != '[':
        print('Not Balanced!')
        break

      if i == '}' and top != '{':
        print('Not Balanced!')
        break

  if not stack:
    print('Balanced!')
else:
  print('Not Balanced!')
```

```python
# 4: Shopping Cart System
class ShoppingCart:
  def __init__(self):
    self.items = {}
  def add_item(self, name, price, quantity):
    if quantity <=0 or price < 0:
      print('Invalid quantity or price')
      return

    if name in self.items:
      self.items[name]['quantity'] += quantity

    else:
      self.items[name] = {'price' : price, 'quantity' : quantity}

  def remove_items(self, name):
    if name in self.items:
      del self.items[name]
      print(f"{name} removed!")
    else:
      print(f"{name} doesn't exists")

  def calculate_total(self):
    total = 0
    for i in self.items.values():
      total += i['price'] * i['quantity']

    return total

  def display_cart(self):
    if not self.items:
      print('Cart is empty')
      return

    for name, details in self.items.items():
      subtotal = details['price'] * details['quantity']
      print(f"{name} - Qty:{details['quantity']} - Subtotal:{subtotal}")

    print(f"Total: {self.calculate_total()}")

cart = ShoppingCart()

cart.add_item("Keyboard", 1500, 2)
cart.add_item("Mouse", 500, 1)
cart.add_item("Keyboard", 1500, 1)

cart.display_cart()
```

```python
# 5: Analyze Monthly Expenses Using Pandas
import pandas as pd

data = {
    "Category": [
        "Groceries", "Transport", "Groceries",
        "Utilities", "Transport", "Utilities"
    ],
    "Month": [
        "January", "January", "February",
        "January", "February", "February"
    ],
    "Amount": [4500, 1200, 3800, 1800, 1500, 2000]
}

df = pd.DataFrame(data)
category_total = df.groupby('Category')['Amount'].sum()
months_total = df.groupby('Month')['Amount'].sum()
highest_category = category_total.idxmax()
high_expenses = df[df['Amount'] > 1500]
df['BudgetStatus'] = df['Amount'].apply(lambda amount: 'High' if amount>2000 else 'Normal')

print(df)
print()
print('Category_total:-')
print(category_total)
print()
print('Month_total:-')
print(months_total)
print()
print("Highest spending category:",highest_category)
print()
print('Expenses greater than 1500:', high_expenses)
print()
print("Expenses with budget status:")
print(df)
```
