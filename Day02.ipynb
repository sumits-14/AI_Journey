```python
# 3: Sentence Inspector
sentence = "An intelligent engineer builds useful applications"
print("Total words:",len(sentence.split()))
l = 0
w = ''
for i in sentence.split():
  if len(i) > l:
    l = len(i)
    w = i
print("Longest word:",w)
print(sentence.upper())
c = 0
for i in sentence.lower().split():
  if i.startswith('a'):
    c+=1
print('Words starting with a vowel:',c)
```

```python
# 4: Library Book Management
class LibraryBook:
  def __init__(self, title, author):
    self.title = title
    self.author = author
    self.is_borrowed = False

  def display_info(self):
    print(f"Book: {self.title}")
    print(f"Author: {self.author}")

    if self.is_borrowed:
      print("Status: Borrowed")
    else:
      print("Status: Available")

  def borrow_book(self):
    if self.is_borrowed:
      print(f"{self.title} is already borrowed.")
    else:
      self.is_borrowed = True
      print(f"{self.title} has been borrowed successfully.")

  def return_book(self):
    if self.is_borrowed:
      self.is_borrowed = False
      print(f"{self.title} has been returned successfully.")
    else:
      print(f"{self.title} was not borrowed.")

book1 = LibraryBook("Python Basics", "JS")
book2 = LibraryBook('DS Basic', "JD")

book1.display_info()
print()

book1.borrow_book()
book1.borrow_book()
book1.return_book()
book1.return_book()
print()

book2.display_info()
```

```python
# 5: Normalize Data Using NumPy
import numpy as np

data = np.array([10, 20, 30, 40, 50])

max = np.max(data)
min = np.min(data)

if max != min:
  normalize_data = (data-min) / (max - min)
else:
  normalize_data = np.zeros_like(data, dtype=float)

print("Original Array:", data)
print(normalize_data)
print("Normalized min:", np.min(normalize_data))
print("Normalized max:", np.max(normalize_data))
```
