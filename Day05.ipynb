```python
# 1: Move All Zeros to the End
numbers = [0, 5, 0, 3, 8, 0, 2]
z = []
for i in numbers:
  if i == 0:
    z.append(i)
    numbers.remove(i)
print(numbers+z)
```

```python
numbers = [0, 5, 0, 3, 8, 0, 2]
nz = [n for n in numbers if n!=0]
z = [n for n in numbers if n==0]
print(nz+z)
```

```python
# 2: Find the First Non-Repeating Character
# text = 'swiss'
# text = "aaBBcc"
# text = 'ajjsa hdkasjweo'
text = 'aabbkkkab'
text = text.lower()
fr = {}
for i in text:
  if i !=' ':
    if i in fr:
      fr[i] += 1
    else:
      fr[i] = 1
for k in fr:
  if fr[k] == 1:
    print(k)
    break

else:
  print('No non-repeating character')
```

```python
# 3: Sort Employees by Salary
employees = [
    {"name": "Rahul", "salary": 45000},
    {"name": "Priya", "salary": 60000},
    {"name": "Amit", "salary": 45000},
    {"name": "Sneha", "salary": 75000}
]
se = sorted(employees, key=lambda p : (-p['salary'], p['name']))
for e in se:
  print(f"{e['name']} - {e['salary']}")
```

```python
# 4: Parking Lot Management System
class ParkingLot:
  def __init__(self, capacity):
    self.capacity = capacity
    self.parked_vehicals = []

  def park_vehicle(self, registration_number):
    if len(self.parked_vehicals) < self.capacity:
      self.parked_vehicals.append(registration_number)
      print(f"{registration_number} parked successfully.")
    else:
      print("Parking lot is full.")

  def remove_vehicle(self, registration_number):
    if registration_number in self.parked_vehicals:
      self.parked_vehicals.remove(registration_number)
      print(f"{registration_number} removed successfully.")
    else:
      print(f"{registration_number} is not parked here.")

  def display_vehicles(self):
    print("Parked vehicals:")
    if len(self.parked_vehicals) == 0:
      print(f"No car is parked.")
    else:
      for i in self.parked_vehicals:
        print(i)

  def available_spaces(self):
    return self.capacity - len(self.parked_vehicals)

parking = ParkingLot(2)

parking.park_vehicle("MH12AB1234")
parking.park_vehicle("MH14CD5678")
parking.park_vehicle("MH12EF9012")
print()

parking.display_vehicles()
print()

print("Available spaces:", parking.available_spaces())
parking.remove_vehicle("MH12AB1234")
print("Available spaces:", parking.available_spaces())
```

```python
# 5: Analyze Student Performance Using Pandas
import pandas as pd

data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Vijay"],
    "Class": ["A", "A", "B", "B", "A"],
    "Maths": [85, 92, 45, 78, 60],
    "Science": [80, 88, 50, 82, 65],
    "English": [75, 95, 40, 85, 70]
}
df = pd.DataFrame(data)

subjects = ['Maths', 'Science', 'English']
df['Average'] = df[subjects].mean(axis=1)

df['Result'] = df['Average'].apply(lambda avg: 'Pass' if avg >=60 else 'Fail')

print('Student averages:')
print(df[['Class','Name', 'Average']].to_string(index=False, formatters={'Average': "{:.2f}".format}))

print('\nResults:')
print(df[['Name', "Result"]].to_string(index=False))

print("\nStudents with average > 80:")
print(df.loc[df['Average'] > 80, 'Name'].to_string(index=False))

df['SubjectAverage'] = df[subjects].mean(axis=1)
class_averages = df.groupby("Class")["SubjectAverage"].mean()
print("\n",class_averages)
print(df['SubjectAverage'])

print("\nClass-wise average:")
for class_name, avg in class_averages.items():
  print(f"Class {class_name}: {avg: .2f}")

top_index = df['Average'].idxmax()
print("\nTop performer:", df.loc[top_index, 'Name'])
```
