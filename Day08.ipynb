```python
# 3: Convert Roman Numerals to Integers
def roman_to_int(l):
  rd = {'I' : 1,
  'V' : 5,
  'X' : 10,
  'L' : 50,
  'C' : 100,
  'D' : 500,
  'M' : 1000}

  l = l[::-1]
  sum = rd[l[0]]
  for i in range(1, len(l)):
    if rd[l[i - 1]] > rd[l[i]]:
      sum -= rd[l[i]]
    else:
      sum += rd[l[i]]

  return sum

# print(roman_to_int("MCMXCIV"))
# print(roman_to_int("LVIII"))
print(roman_to_int("IX"))
```
