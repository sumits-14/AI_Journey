```python
# 2: Find duplicate elements
l1 = [1, 2, 3, 2, 4, 3, 3]
l = []
j = []
for i in l1:
  if i not in l:
    l.append(i)
  else:
    j.append(i)
list(set(j))
```

```python
# 3: Character frequency
st1 = "Hello World"
fr = {}
for i in st1:
  if i in fr:
    fr[i] += 1
  else:
    fr[i] = 1
fr
```

```python
# Create a BankAccount class that stores a balance and provides
# deposit(), withdraw() and get_balance() methods. Deposits must
# be positive, and withdrawals must not exceed the available balance.

class BankAccount:
  def __init__(self, balance):
    self.__balance = balance

  @property
  def totalBalance(self):
    print('Total Balance is : ', self.__balance)

  def deposit(self, amount):
    if amount>0:
      self.__balance += amount
    else:
      print("Sorry! Can't procedd this request")

  def withdraw(self, amount):
    if amount<self.__balance:
      self.__balance -= amount

acc = BankAccount(1000)
acc.totalBalance
acc.withdraw(300)
acc.totalBalance
```

```python
# Create a NumPy array containing integers from 1 through 20.
# Calculate its mean (average), median (middle value), minimum,
# maximum and standard deviation (how spread out the values are).
import numpy as np
a = np.array(np.arange(1,21))
print(a)
print(np.mean(a))
print(np.median(a))
print(np.min(a))
print(np.max(a))
print(np.std(a))
```
