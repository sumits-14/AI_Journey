```python
# 1: Password Strength Checker
pswrd = input("Enter your password:")
u = 0
l = 0
d = 0
s = 0
if len(pswrd) >= 8:
  for i in pswrd:
    if i.isupper():
      u = 1
    if i.islower():
      l = 1
    if i.isdigit():
      d = 1
    if i in '@#!':
      s = 1
if u and l and d and s and len(pswrd)>=8:
  print('Password strength: Strong')
else:
  print('Reenter password!')
```

```python
# 2: Find the Second-Largest Unique Number
numbers = [88, 12, 45, 7, 45, 32, 52, 12, 28, 74, 99, 98, 100]
l = 0
sl = 0
for i in numbers:
  if i > l:
    sl = l
    l = i
  if i < l and i > sl:
    sl = i
# print(l)
print(f"Second-largest unique number: {sl}")
```

```python
# numbers = [-8, -3, -12, -3]
l = None
sl = None

for i in numbers:
  if l is None or i > l:
    if i != l:
      sl = l
      l = i
  elif i != l and (sl is None or i > sl):
    sl = i

if sl is None:
  print("There is no second largest number")
else:
  print("Second largest number is:", sl)
```

```python
# 3: Word Frequency Ranking
import string
sentence = "Python is fun, and Python is powerful!"
# mt = str.maketrans("", "", string.punctuation)    // Both the steps are correct but this one need import
mt = str.maketrans("", "", ".,!@#$%&*_-+=")
tr = sentence.translate(mt)
fr = {}
for i in tr.split():
  if i in fr:
    fr[i] += 1
  else:
    fr[i] = 1
print(fr)
```

```python

class Employee:
  def __init__(self,name, basic_salary, bonus_percentage):
    self.name = name
    self.basic_salary = basic_salary
    self.bonus_percentage = bonus_percentage

  def calculate_bonus(self):
    return self.basic_salary*self.bonus_percentage

  def calculate_annua_salary(self):
    return (self.basic_salary+self.calculate_bonus())*12

  def display_details(self):
    print("Employee:", self.name)
    print("Basic Monthly Salary:", self.basic_salary)
    print("Monthly Bonus:", self.calculate_bonus())
    print("Annual Salary:", self.calculate_annua_salary())

salary = Employee('Pratik', 30000, 0.10)
salary.display_details()
```

```python
import numpy as np

# 5: NumPy Array Statistics
scores = np.array([65, 80, 75, 90, 55, 85, 70, 95])
c = 0
m = []
for i in scores:
  if i >= 80:
    c+=1
  if i > np.mean(scores):
    m.append(i)
print("Mean:",np.mean(scores))
print("Median:", np.median(scores))
print("Standard deviation:", np.std(scores))
print("Students scoring at least 80:", c)
# print("Scores at or above mean:", [n for n in scores if n > np.mean(scores)])
print("Scores at or above mean:", m)
```
